# Titanic Dataset Profiling with Pandas

This beginner-friendly notebook explores the Titanic dataset with Pandas and creates an automated data profiling report.

## Project Introduction

Pandas is used to load, inspect, and summarize tabular data. YData Profiling builds a detailed HTML report that helps show data types, missing values, distributions, correlations, and duplicate observations.

The goal of this project is to practice basic data profiling rather than build a machine-learning model.

## 3. Importing Required Libraries

`pandas` provides DataFrame operations for reading and exploring tabular data. The standard-library `pathlib.Path` class is used to build reliable, platform-independent dataset paths.

````python
from pathlib import Path

import pandas as pd
````

## 4. Loading the Titanic Dataset

`read_csv()` loads a comma-separated values file into a Pandas DataFrame. The code below searches for `data/titanic.csv` when the notebook is run from either the project directory or the `notebooks` directory.

````python
project_root = Path.cwd().resolve()

if (project_root / "data" / "titanic.csv").exists():
    data_path = project_root / "data" / "titanic.csv"
elif (project_root.parent / "data" / "titanic.csv").exists():
    project_root = project_root.parent
    data_path = project_root / "data" / "titanic.csv"
else:
    raise FileNotFoundError(
        "The original Titanic CSV was not found. Place it at data/titanic.csv "
        "before running this notebook."
    )

df = pd.read_csv(data_path)
print(f"Loaded dataset: {data_path}")
````

## 5. Exploring the Dataset

`shape` reports the number of rows and columns. `info()` summarizes column names, data types, and non-null values.

````python
print("Dataset shape:", df.shape)
df.info()
````

## 6. Viewing the First Five Rows

`head()` displays the first five rows by default. It is a quick way to verify that the CSV loaded as expected.

````python
df.head()
````

## 7. Viewing the Last Five Rows

`tail()` displays the final five rows. Checking both ends can reveal unexpected formatting or placeholder values.

````python
df.tail()
````

## 8. Statistical Summary

`describe()` generates summary statistics for numeric columns, including count, mean, standard deviation, minimum, quartiles, and maximum.

````python
df.describe()
````

## 10. Generating the Data Profiling Report

`ProfileReport` analyzes the DataFrame and prepares the interactive profiling report. `to_file()` exports that report as an HTML file.

````python
from pathlib import Path

reports_dir = project_root / "reports"
reports_dir.mkdir(parents=True, exist_ok=True)

profile = ProfileReport(
    df,
    title="Titanic Dataset Profiling Report",
)

report_path = reports_dir / "titanic_profiling_report.html"
profile.to_file(report_path)
print(f"Profiling report saved to: {report_path}")
````
